# Data Loading (T2.5)

Loads the source data into the 19 tables of the 3NF schema in DBRepo via the REST API (`dbrepo` Python client). Each lookup table and `data_guide` has its own CSV. The `collisions` fact table is loaded in 10 chunks from the main DfT CSV.

## Setup

Sets up the `dbrepo` REST client and pandas. Username and password are entered at runtime, not stored in the file.

```python
import pandas as pd
from dbrepo.RestClient import RestClient
from getpass import getpass

database_id = "82c19b39-246c-4409-b25c-8baf3a158a70"

USERNAME = input("DBRepo username: ")
PASSWORD = getpass("DBRepo password: ")

client = RestClient(
    endpoint="https://test.dbrepo.tuwien.ac.at",
    username=USERNAME,
    password=PASSWORD,
)
```
